importing the Dependencies

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

Data Collection and processing

```python
#loading the data from csv file to pandas dataframe
car_dataset=pd.read_csv('/content/car data.csv')
```

```python
print(car_dataset.head())
#inspecting the first 5 rows of the dataframe
```

```python
#checking the number of missing values
car_dataset.isnull().sum()
```

```python
#checking the distribution of categorical data
print(car_dataset.Fuel_Type.value_counts())
print(car_dataset.Seller_Type.value_counts())
print(car_dataset.Transmission.value_counts())
```

```python
#encoding the categorical data
#encoding 'Fuel_Type' Column
car_dataset.replace({'Fuel_Type':{'Petrol':0,'Diesel':1,'CNG':2}},inplace=True)
car_dataset.replace({'Seller_Type':{'Dealer':0,'Individual':1}},inplace=True)
car_dataset.replace({'Transmission':{'Manual':0,'Automatic':1}},inplace=True)
```

```python
car_dataset
```

Assigning the labels and features--I/P and O/P

```python
X=car_dataset.drop({'Selling_Price','Car_Name'},axis=1)
Y=car_dataset['Selling_Price']
```

```python
#plt.scatter(X,Y)
print(X)
print(Y)
```

Splitting the data into training and test data

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=.1,random_state=2)
```

```python
print(X_train.shape,X_test.shape,Y_train.shape,Y_test.shape)
```

ModelTraining
> LinearRegression Model



```python
#Loading LinearRegression Model
lin_reg_model=LinearRegression()
```

```python
lin_reg_model.fit(X_train,Y_train)
```

Model Evaluation

```python
#predicting on training data
training_data_prediction=lin_reg_model.predict(X_train)
```

```python
#R2_score,Mean2_Error
error_score=metrics.r2_score(Y_train,training_data_prediction)
print('R squared error: ', error_score);
```

```python
error_score=metrics.r2_score(training_data_prediction,Y_train)
print('R squared error: ', error_score);
```
