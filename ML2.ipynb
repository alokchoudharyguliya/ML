```python
import pandas as pn
import numpy as nm
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plo
```

```python
#inserting data into our program
train=pn.read_csv('train (4).csv')
test=pn.read_csv('test (1).csv')
```

```python
#assigning x and y their data frames
x=train.drop(columns='Y')
y=train['Y']
print(y.shape,x.shape)
```

```python
#plotting graph for training data
plo.scatter(x,y)
plo.xlabel('feature')
plo.ylabel('labels')
```

```python
#using LinearRegression algorithm to train model
model=LinearRegression()
```

```python
#training model
model.fit(x,y)
```

```python
x_test=test.drop(columns='Y')
y_test=test['Y']

#plotting testing data
plo.xlabel('Test_feature')
plo.ylabel('Test_labels')
plo.scatter(x_test,y_test)
```

```python
y_predict=model.predict(x_test)
plo.scatter(x_test,y_test,color='green',linewidths=9)
plo.scatter(x_test,y_predict,color='red',linewidths=1)
```

```python
r2_score(x_test,y_predict)
```
