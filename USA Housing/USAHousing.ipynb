```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split
data=pd.read_csv('USA_Housing.csv')
data
data['Garden']=np.random.choice(['Yes','No'],size=len(data.T.columns))
data.shape
data.head()
# data.tail()
data.columns
type(data.columns)
for i in data.columns:
    print(i)
data.describe()
data.describe().T
data.info()
data.describe(include='object')
data.nunique()
data
data['Price'].nunique
numerical_col=data.select_dtypes(include=np.number).columns
for col in numerical_col:
    print(data[col].value_counts())
    print("-------------------------------------\n-------------------------------------\n")
categorical_col=data.select_dtypes(include='object').columns
for col in categorical_col:
    print(data[col].value_counts())
    print("-------------------------------------\n-------------------------------------\n")
# for col in categorical_col:
#     sns.countplot(data=data,x=col)
#     plt.grid()
#     plt.show()
# Creating a function to do all the tasks
def attribute_inof(df_feature):
    print("Description of attribute",'\n')
    print(df_feature.describe(),'\n')
attribute_inof(data)
for column in data:
    print('\033[1m'+'Attribute_Name :', column+'\033[0m')
## Data Visualization
data
data.hist(figsize=(15,12))
data.duplicated().sum()
sns.boxplot(data['Price'])
plt.show()
data
sns.boxplot(data['Avg. Area Income'])
plt.show()
data.shape
data[data['Price']>2200000].index
# data.drop()
numerical_col = data.select_dtypes(include=np.number).columns
plt.figure(figsize=(15,10))
for i, col in enumerate(numerical_col):
    plt.subplot(3, 3, i+1)
    sns.boxplot(data[col])
    plt.title(col)
plt.tight_layout()
sns.pairplot(data)
plt.show()
numerical_col = data.select_dtypes(include=np.number).columns
plt.figure(figsize=(15,10))
for i, col in enumerate(numerical_col):
    plt.subplot(3, 3, i+1)
    sns.scatterplot(x=col,y='Price',data=data)
    plt.title(col)
plt.tight_layout()
# numerical_col = data.select_dtypes(include='object').columns
# plt.figure(figsize=(15,10))
# for i, col in enumerate(numerical_col):
#     plt.subplot(3, 3, i+1)
#     sns.scatterplot(x=col,y='Price',data=data)
#     plt.title(col)
# plt.tight_layout()
## Checking correlation
data.columns
df=data[['Avg. Area Income','Area Population','Avg. Area House Age','Avg. Area Number of Bedrooms','Avg. Area Number of Rooms','Price']]
df
df.corr()
corr=df.corr()
plt.figure(figsize=(5,5))
sns.heatmap(corr,annot=True)
plt.show()
def correlation(df,threshold):
    cor=set()
    cor_matrix=df.corr()
    for i in range(len(cor_matrix.columns)):
        for j in range(i):
#             print(i,j)
            if abs(cor_matrix.iloc[i,j])>threshold:
                print(cor_matrix.iloc[i,j])
                colname=cor_matrix.columns[i]
                print(colname)
                cor.add(colname)
    return cor
cor_features=correlation(df,0.4)
print(len(cor_features))
cor_features
len(df.columns)
df.corr()['Price']
## mapping categorical values to some numbers: Encoding
data['Garden']=data['Garden'].map({'Yes':1,'No':0})
data.corr()
corr=data.corr()
plt.figure(figsize=(15,10))
sns.heatmap(corr,annot=True)
plt.show()
cor_features=correlation(data,0.4)
print(len(cor_features))
cor_features
data.corr()['Price']
## Model building
data
data.drop('Address',axis=1,inplace=True)
data.iloc[5,:]
data.iloc[:,5]
data['Price']
data.columns
X=data.drop('Price',axis=1)
# axis 1 : Columns
# axie 0 : Rows
X
y=data['Price']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=0,test_size=0.2)
X_train
X_test
y_train
y_test
X_train.shape
y_train.shape
y_test.shape
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
X_train_pred=model.predict(X_train)
X_test_pred=model.predict(X_test)
len(X_train_pred)
coeff_df=pd.DataFrame(model.coef_,X.columns,columns=['Coefficient'])
coeff_df
plt.figure(figsize=(18,8))
plt.scatter(y_train,X_train_pred)
plt.figure(figsize=(18,8))
plt.scatter(y_test,X_test_pred)
plt.plot([y_test.min(),y_test.max()],[y_test.min(),y_test.max()],'k--',linewidth=5)
## For training data
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
mse=mean_squared_error(y_train,X_train_pred)
mse
rmse=np.sqrt(mse)
rmse
ame=mean_absolute_error(y_train,X_train_pred)
ame
r2=r2_score(y_train,X_train_pred)
r2
## for test data

mset=mean_squared_error(y_test,X_test_pred)
print(mset)
rmset=np.sqrt(mset)
print(rmse)
amet=mean_absolute_error(y_test,X_test_pred)
print(ame)
r2t=r2_score(y_test,X_test_pred)
print(r2t)

print('Training Set Evaluation: ', '\n________________________________')
print('MSE: ', mse)
print('RMSE: ', rmse)
print('MAE: ', ame)
print('R2_Score: ', r2)
print('Test Set Evaluation: ', '\n________________________________')
print('MSE: ', mset)
print('RMSE: ', rmset)
print('MAE: ', amet)
print('R2_Score: ', r2t)
```
