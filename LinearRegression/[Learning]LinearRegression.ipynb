```python
import pandas as po #for handling data as data frame
import numpy as nm #for using arrays
from sklearn.linear_model import LinearRegression #Model for ML
from sklearn.metrics import accuracy_score #To determine accuracy of model
from sklearn.metrics import r2_score
import matplotlib.pyplot as ploot # for data visualiszation
```

Importing data into a pandas data frame

```python
#Import data into train and test dataframes
train=po.read_csv("/content/train (1).csv")
test=po.read_csv('/content/test.csv')
train.head()
test.head()
train.dropna()
test.dropna()
#y is dependent variable called as labels
#x is independent variable called as features
```

```python
print(train.shape,test.shape)
```

```python
x=train.drop(columns='y')
#x=train['x']
y=train['y']
#y=train.drop(columns='x')
print(type(y))
#print(y['x'][3])
```

```python
ploot.scatter(x,y)
ploot.xlabel('Features')
ploot.ylabel('Labels')
```

import model estimation class to variable

```python
model=LinearRegression()
#using algorithm
```

training model

```python
x = nm.array(x)
Y = nm.array(y)
model.fit(x,Y)
#training done
```

```python
print(type(y))
```

```python
x_test=test.drop(columns='y')
y_test=test['y']
ploot.xlabel('Test_Features')
ploot.ylabel('Test_Labels')
ploot.scatter(x_test,y_test)
```

```python
y_predict=model.predict(x_test)
#y_predictarr=nm.array(y_predict)
ploot.scatter(x_test,y_test)
ploot.scatter(x_test,y_predict)
print(type(x_test),type(y_predict))
```

Accuracy_score

```python
r2_score(x_test,y_predict)
```

**R2_score**
0->model gives mean of data
1->Best score
-ve->very high deviation
