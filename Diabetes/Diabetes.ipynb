```python
#====IMPORTING THE DEPENEDENCIES======
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler #standardizing the data
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
#=====DATA COLLECTION ANALYSIS======
#PIMA Diabetes dataset
diabetes_dataset=pd.read_csv('diabetes.csv')
```

```python
diabetes_dataset.groupby('Outcome').mean()
```

```python
#separating the dataset as input and output
x=diabetes_dataset.drop(columns='Outcome',axis=1)
Y=diabetes_dataset['Outcome']
```

```python
print(x)
```

```python
print(Y)
```

```python
#Data_Standardization
scaler=StandardScaler()
```

```python
scaler.fit(x)
```

```python
standardized_data=scaler.transform(x)
```

```python
print(standardized_data)
```

```python
X=standardized_data
Y=diabetes_dataset['Outcome']
```

```python
#Train_Test_Split
#test_size=gives fraction of data as training data
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(x.shape,X_train.shape,X_test.shape)
print(Y.shape,Y_train.shape,Y_test.shape)
```

```python
#training the model
#inserting the svm
#support vector machine
classifier=svm.SVC(kernel='linear')
```

```python
#fitting the data
classifier.fit(X_train,Y_train)
```

```python
#evaluation process-to check how many times our model is classifying accurately
#accuracy_score on the training data
X_train_prediction=classifier.predict(X_train)
#training ddta accuracy finding
training_data_accuracy=accuracy_score(X_train_prediction,Y_train)
```

```python
print(training_data_accuracy)
```

```python
X_test_prediction=classifier.predict(X_test)
test_data_accuracy=accuracy_score(X_test_prediction,Y_test)
```

```python
print(test_data_accuracy)
```

```python
#Making Prediction System
input_data=(1,103,30,38,83,43.3,0.183,33)
#changing the input data to numpy array
input_data_as_Narray=np.array(input_data)
#reshapning the array
input_data_reshaped=input_data_as_Narray.reshape(1,-1)
#standardizing the input data
std_data=scaler.transform(input_data_reshaped)
print(std_data)
```
